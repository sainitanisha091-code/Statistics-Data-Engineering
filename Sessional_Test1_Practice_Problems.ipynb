## Question 1 - Average Daily Sales

A shop noted how many items it sold on each day. The daily sales are given on ONE line, separated by spaces.
(a) Give the average number of items sold per day (add all and share equally), rounded to 2 decimals.
(b) On how many days did the shop sell MORE than that average?
Input: one line of space-separated numbers.

```python
import numpy as np

sales = np.array(input().split(), dtype=float)

avg = sales.mean()                       # (a) average
print(round(avg, 2))
print(int((sales > avg).sum()))          # (b) how many days above average
```

## Question 2 - Middle Temperature

Daily temperatures are given on ONE line separated by spaces.
(a) Give the middle value when they are ordered (equal count -> average of the two middle ones), rounded to 2 decimals.
(b) How many days were WARMER than that middle value?
Input: one line of numbers.  

```python
import numpy as np

temp = np.array(input().split(), dtype=float)

middle = np.median(temp)                 # (a) middle value
print(round(middle, 2))
print(int((temp > middle).sum()))        # (b) how many are warmer
```

## Question 3 - Most Common Rating

Customer star ratings (1 to 5) are given on ONE line separated by spaces.
(a) Which rating appears MOST often? (If tie, print the smaller one.)
(b) How many times was that rating given?
Input: one line of whole numbers.  Output: line 1 = (a), line 2 = (b).

```python
import pandas as pd

rating = pd.Series(input().split(), dtype=int)

top = rating.mode()[0]                   # (a) most common (smallest if tie)
print(top)
print(int((rating == top).sum()))        # (b) how many times
```

## Question 4 - Spread of Prices

Product prices are given on ONE line separated by spaces.
(a) By how much does the dearest cost more than the cheapest? Round to 2 decimals.
(b) What value is exactly halfway between the cheapest and the dearest? Round to 2 decimals.
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

price = np.array(input().split(), dtype=float)

print(round(price.max() - price.min(), 2))        # (a) range
print(round((price.max() + price.min()) / 2, 2))  # (b) mid value
```

## Question 5 - Spread of Marks

Exam marks of a class are given on ONE line separated by spaces.
(a) Find the class average, then the average of the squared distances of each mark from it (divide by the number of marks). Round to 2 decimals.
(b) Give the square root of part (a). Round to 2 decimals.
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

marks = np.array(input().split(), dtype=float)

print(round(np.var(marks), 2))           # (a) average of squared distances (divide by n)
print(round(np.std(marks), 2))           # (b) its square root
```

## Question 6 - Steady Growth Factor

Monthly growth factors (like 1.2 = 20% up) are given on ONE line, all positive.
(a) Find the single constant factor giving the same total growth from first to last month. Round to 2 decimals.
(b) Turn it into percent per month (factor - 1, times 100). Round to 2 decimals.
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

factors = np.array(input().split(), dtype=float)

steady = np.exp(np.mean(np.log(factors)))    # (a) geometric mean via logs
print(round(steady, 2))
print(round((steady - 1) * 100, 2))          # (b) as a percentage
```

## Question 7 - Average Speed (equal distances)

Speeds over equal-distance stretches are given on ONE line, all positive.
(a) The single constant speed giving the same total trip time. Round to 2 decimals. (Simply averaging the speeds is NOT correct here.)
(b) For comparison, the ordinary average of the speeds. Round to 2 decimals.
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

speed = np.array(input().split(), dtype=float)
n = len(speed)

print(round(n / np.sum(1.0 / speed), 2))     # (a) harmonic mean
print(round(np.mean(speed), 2))              # (b) ordinary average
```

## Question 8 - Grade with Credits

TWO lines. Line 1 = subject scores. Line 2 = the credit weight of each subject (same count).
(a) Overall grade where each subject counts in proportion to its credits. Round to 2 decimals.
(b) Plain average that ignores credits. Round to 2 decimals.
Input: line 1 = scores, line 2 = weights.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

score  = np.array(input().split(), dtype=float)
weight = np.array(input().split(), dtype=float)

print(round(np.average(score, weights=weight), 2))   # (a) weighted mean
print(round(np.mean(score), 2))                      # (b) plain mean
```

## Question 9 - Two Kinds of Variance

Rainfall readings are given on ONE line separated by spaces. For both parts add up the squared distances of each value from the average.
(a) Divide that total by the number of readings. Round to 2 decimals.
(b) Divide that same total by ONE LESS than the number of readings. Round to 2 decimals.
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

vals = np.array(input().split(), dtype=float)

print(round(np.var(vals), 2))            # (a) divide by n
print(round(np.var(vals, ddof=1), 2))    # (b) divide by n - 1
```

## Question 11 - Quartiles and Middle Half

House sizes are given on ONE line separated by spaces.
(a) The size below which the smallest 25% fall. Round to 2 decimals.

## Question 12 - Find the Unusual Days

Daily visitor counts are given on ONE line. Take the 25% mark and 75% mark, let W be the gap; the normal range is (25% mark - 1.5*W) up to (75% mark + 1.5*W).
(a) The list of unusual values (outside the range), sorted (empty list [] if none).
(b) The largest count still counted as normal (the upper limit). Round to 2 decimals.
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

vals = np.array(input().split(), dtype=float)
q1 = np.percentile(vals, 25)
q3 = np.percentile(vals, 75)
w = q3 - q1
low, high = q1 - 1.5 * w, q3 + 1.5 * w

print(sorted(vals[(vals < low) | (vals > high)].tolist()))   # (a) outliers
print(round(high, 2))                                        # (b) upper limit
```

## Question 13 - Count the Unusual Days

Daily support tickets are given on ONE line. Use the same normal range: below (25% mark - 1.5*W) or above (75% mark + 1.5*W) is unusual, where W is the gap between the 75% and 25% marks.
(a) How many values are unusual?
(b) How many values are within the normal range?
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

vals = np.array(input().split(), dtype=float)
q1 = np.percentile(vals, 25)
q3 = np.percentile(vals, 75)
w = q3 - q1
outside = (vals < q1 - 1.5 * w) | (vals > q3 + 1.5 * w)

print(int(outside.sum()))                # (a) unusual count
print(int((~outside).sum()))             # (b) normal count
```

## Question 14 - Turn Values into Z-scores

Test scores are given on ONE line. For each score compute (score - average) divided by the spread (population: divide squared distances by the count, then square root).
(a) The list of these z-scores in the ORIGINAL order, each rounded to 2 decimals.
(b) The original score that is furthest from the average.
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

vals = np.array(input().split(), dtype=float)
m = np.mean(vals)
s = np.std(vals)
z = (vals - m) / s

print([round(float(v), 2) for v in z])       # (a) z-scores
print(vals[int(np.argmax(np.abs(z)))])        # (b) furthest value
```

## Question 15 - Unusual by Z-score

Reaction times are given on ONE line. A value is unusual if it is more than 2 spreads from the average (population spread).
(a) The list of unusual values, sorted (empty list [] if none).
(b) How many values are unusual?
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

vals = np.array(input().split(), dtype=float)
m = np.mean(vals)
s = np.std(vals)
d = np.abs((vals - m) / s)

print(sorted(vals[d > 2].tolist()))      # (a) values more than 2 spreads away
print(int((d > 2).sum()))                # (b) how many
```

## Question 16 - Clean, then Average

Daily electricity usage is given on ONE line. Faulty readings sit outside the normal range: below (25% mark - 1.5*W) or above (75% mark + 1.5*W), where W is the gap between the 75% and 25% marks.
(a) Remove the faulty readings and give the average of what remains. Round to 2 decimals.
(b) The average when nothing is removed. Round to 2 decimals.
Input: one line of numbers.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

vals = np.array(input().split(), dtype=float)
q1 = np.percentile(vals, 25)
q3 = np.percentile(vals, 75)
w = q3 - q1
low, high = q1 - 1.5 * w, q3 + 1.5 * w
kept = vals[(vals >= low) & (vals <= high)]

print(round(np.mean(kept), 2))           # (a) average after removing faulty
print(round(np.mean(vals), 2))           # (b) average of everything
```

## Question 17 - Sales by Region

TWO lines. Line 1 = region names (words). Line 2 = the sales for each row (same count).
(a) The average sales for each region, one line per region as  region: value  (2 decimals), regions in alphabetical order.
(b) A final line  Highest: <region>  for the region with the largest average.
Input: line 1 = regions, line 2 = sales.

```python
import pandas as pd
import numpy as np

regions = input().split()
sales = np.array(input().split(), dtype=float)
df = pd.DataFrame({"region": regions, "sales": sales})

per = df.groupby("region")["sales"].mean().sort_index()
for region, avg in per.items():
    print(region + ": " + str(round(avg, 2)))
print("Highest: " + per.idxmax())
```

## Question 18 - Hours vs Scores

TWO lines. Line 1 = hours studied. Line 2 = exam score for each student (same count).
(a) A number between -1 and +1 for how strongly they move together. Round to 2 decimals.
(b) Print Positive if it is above 0, Negative if below 0, or None if exactly 0.
Input: line 1 = hours, line 2 = scores.  Output: line 1 = (a), line 2 = (b).

```python
import numpy as np

hours = np.array(input().split(), dtype=float)
score = np.array(input().split(), dtype=float)

c = np.corrcoef(hours, score)[0, 1]
print(round(c, 2))                                          # (a) correlation
print("Positive" if c > 0 else ("Negative" if c < 0 else "None"))   # (b) direction
```

## Question 19 - Quick Summary of Ages

Employee ages are given on ONE line separated by spaces. Treat them as a SAMPLE.
Print exactly five lines in this order and format:
Count: <how many>
Min: <smallest, 2 decimals>
Max: <largest, 2 decimals>
Mean: <average, 2 decimals>
Std: <sample spread (divide by count-1), 2 decimals>

## Question 20 - Missing Readings

Sensor readings are given on ONE line separated by spaces. When the sensor failed the value is written as NA.
(a) How many readings are missing (how many NA)?
(b) The average of the readings that ARE present. Round to 2 decimals.
Input: one line of numbers and NA tokens.  Output: line 1 = (a), line 2 = (b).

```python
import pandas as pd

nums = pd.to_numeric(pd.Series(input().split()), errors="coerce")   # NA -> missing

print(int(nums.isna().sum()))            # (a) how many missing
print(round(nums.mean(), 2))             # (b) average of the present ones
```
